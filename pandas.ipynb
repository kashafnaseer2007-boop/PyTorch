###Pandas Series: The Building Block of Data Analysis

**A Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floats, Python objects, etc.).**

It has two main components:


*   values: The actual data (like a NumPy array)
*   index: Labels for each element (like row names)


```python
# Creating a Series

import pandas as pd

# Basic series (default integer index)
s = pd.Series([10, 20, 30, 40])
print(s)
```

```python
s
```

```python
# With Custom Index

s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s)
```

```python
#  From a Dictionary
data = {'apple': 5, 'banana': 3, 'cherry': 8}
s = pd.Series(data)
print(s)
```

```python
# From a Scalar (Broadcasting)
s = pd.Series(5, index=['x', 'y', 'z'])
print(s)
```

```python
# Key Properties

s = pd.Series([1, 2, 3], index=['x', 'y', 'z'])
print(s.values)
print(s.index)
```

```python
s
```

```python
# Accessing Elements
# By Position (Integer Location)

s = pd.Series([10, 20, 30, 40,50], index=['a', 'z','b', 'c', 'd'])
print(s)
print(s[0])      # 10 (first element)
print(s.iloc[2]) # 30 (explicit integer location)
```

```python
# By Label (Index)

print(s['b'])    # 20
print(s.loc['c'])# 30 (explicit label-based access)
```

```python
# Slicing
print(s['a':'c'])   # Includes 'c'! → ['a', 'b', 'c']
print(s[0:3])
```

```python
# . Vectorized Operations

import numpy as np
s = pd.Series([1, 2, 3, 4])

# Arithmetic
print(s + 10)    # [11, 12, 13, 14]
print(s * 2)     # [2, 4, 6, 8]

# Math functions
print(s ** 2)    # [1, 4, 9, 16]
print(np.sqrt(s))# [1.0, 1.414, 1.732, 2.0]

# Boolean masking
print(s > 2)     # [False, False, True, True]
print(s[s > 2])  # [3, 4]
```

```python
# Boolean masking
print(s > 2)     # [False, False, True, True]


# s[s>2]
print(s[s > 2])  # [3, 4]
```

```python
# Alignment: The Magic Feature
# When operating on two Series, Pandas aligns by index:


s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])



s2 = pd.Series([10, 20, 30], index=['b', 'c', 'd'])

print(s1 + s2)
```

```python
s = s1 + s2
s
```

```python
s.fillna(0)
```

```python
s = pd.Series([1, 2, np.nan, 4,4])
s
print(s.fillna(0))      # [1, 2, 0, 4]
print(s.value_counts()) # Counts non-null values
```

```python
s
```

```python
dates = pd.date_range('2025-01-01', periods=5)
prices = pd.Series([100, 102, 101, 105, 107], index=dates)

print(prices)


print(prices['2025-01-03'])  # 101
print(prices.pct_change())   # Daily % change
```

###Pandas DataFrame: The Heart of Data Science in Python

**A DataFrame is a 2-dimensional labeled data structure with:**


*   Rows (indexed by labels)
*   Columns (each with a name and data type)
* Can hold heterogeneous data (integers, floats, strings, booleans, etc.)



```python
# Creating a DataFrame
# From a Dictionary

import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Tokyo']
}

df = pd.DataFrame(data)
df
```

```python
# From a List of Dictionaries
data = [
    {'Name': 'Alice', 'Age': 25},
    {'Name': 'Bob', 'Age': 30, 'City': 'London'}
]
df = pd.DataFrame(data)
# Missing values become NaN
df
```

```python
adults = df[df['Age'] > 25]
adults
```

```python
df["Name"]
```

```python
df['Country'] = ['USA', 'UK']
df['Is_Adult'] = df['Age'] >= 18
df
```
