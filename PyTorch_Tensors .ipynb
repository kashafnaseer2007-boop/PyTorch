```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data
print(f"Random Tensor: \n {x_rand} \n")
```

```python
tensor = torch.ones(5, 4)
print(tensor)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:,1] = 0
print(tensor)
```

#  Example in 3D
**t = torch.randn(2, 3, 4)** # shape: (2, 3, 4)

**t[:, :, -1]**    # keep dim0 and dim1, take last index of dim2

**t[..., -1]**  # means: take all dimensions before the last, then index -1

## The ... stands in for the first two dimensions (:, :), so t[..., -1] is exactly the same as t[:, :, -1].

```python
a = torch.tensor([[1, 2],
                  [3, 4]])
print(a.shape)
t1 = torch.cat([a, a, a], dim=0)   # dim=0 -> rows. dim=1 -> columns
print(t1)
print(t1.shape)
```

```python
# This computes the matrix multiplication between two tensors. y1, y2, y3 will have the same value
# ``tensor.T`` returns the transpose of a tensor
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.rand_like(y1)
print(torch.matmul(tensor, tensor.T, out=y3))


# This computes the element-wise product. z1, z2, z3 will have the same value
z1 = tensor * tensor
z2 = tensor.mul(tensor)

z3 = torch.rand_like(tensor)
print(torch.mul(tensor, tensor, out=z3))
```

```python
print(tensor, "\n")
print(y1, "\n")
print(y2, "\n")
print(y3, "\n")
print(z1, "\n")
print(z2, "\n")
print(z3, "\n")
```

# Key Difference

**Matrix multiplication**	@ or .matmul()	,Combines rows and columns (sum of products)

**Element-wise multiplication**	* or .mul()	,Multiplies matching positions only


```python
agg = tensor.sum()
agg_item = agg.item()
print(agg_item, type(agg_item))  # sum process = 15 × 1 + 5 × 0 = 15.0.
```

# Tensor to NumPy array

```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

# NumPy array to Tensor

```python
n = np.ones(5)
print(f"n: {n}")
t = torch.from_numpy(n)
print(f"t: {t}")
```

```python
np.add(n, 1, out=n)   # in-place operation
print(f"t: {t}")
print(f"n: {n}")
```
