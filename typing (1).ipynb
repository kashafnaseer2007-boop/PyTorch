```python
#  Function with Default Arguments & Union Types

from typing import Union

 def greet(name: str, age: Union[int, None] = None) -> str:
    """Generate greeting with optional age."""
    if age is not None:
        return f"Hello {name}, you are {age} years old!"

    return f"Hello {name}!"

# Usage
print(greet("Alice"))        # "Hello Alice!"
print(greet("Bob", 30))      # "Hello Bob, you are 30 years old!"
```

```python
# List/Dict Processing with Generic Types

from typing import List, Dict

def calculate_average(scores: List[int]) -> float:
    """Calculate average of a list of scores."""
    print(scores)

    return sum(scores) / len(scores) if scores else 0.0

def get_user_emails(users: Dict[str, str]) -> List[str]:
    """Extract email values from user dict."""
    return list(users.values())

# Usage
avg = calculate_average([95.5, 88.0, 92.5,99])  # 92.0
emails = get_user_emails({"alice": "a@x.com", "bob": "b@y.com"})  # ["a@x.com", "b@y.com"]
print(avg)
print(emails)
```

```python
# Optional Parameters & Return Types
from typing import Optional

def find_max(numbers: List[int]) -> Optional[int]:
    """Return max value or None if empty list."""
    # hsuidhuisdhius
    # ajsdbjsad

    return max(numbers) if numbers else None

# Usage
print(find_max([1, 5, 3]))  # 5
print(find_max([]))         # None
```
